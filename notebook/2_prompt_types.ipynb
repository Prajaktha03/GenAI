### Prompt Types

```python
prompt=[
    {
        "role": "system",
        "content": "You are python developer",
    },
    {
        "role":"user",
        "content":"What is a  list?"
    }
]
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
llm=ChatGroq(
    model="openai/gpt-oss-120b"
)
```

```python
res=llm.invoke(prompt)
res.content
```

### DYNAMIC PROMPTS

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt=ChatPromptTemplate.from_messages([
    {
        "role":"system",
        "content":"You are a language translator, translate the text into {language} "
    },{
        "role":"user",
        "content":"{query}"
    }
])
```

```python
final_prompt=prompt.format_messages(language="Kannada",query="Hi how are you?")
res=llm.invoke(final_prompt)
res.content
```

```python
res=llm.invoke(prompt.invoke({"language":"Kannada","query":"Hi how are you?"}))
```

### chaining 

```python
chain=prompt | llm
```

```python
res=chain.invoke({"language":"Hinglish","query":"Hi how are you?"})
res.content
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
out=StrOutputParser()
```

```python
out.invoke(res.content)
```

```python
chain=prompt|llm|out
```

```python
res=chain.invoke({"language":"Hinglish","query":"Hi how are you?"})
res
```

```python
#Custom function
def typeconversion(res:str):
    return res.upper()
```

```python
chain=prompt | llm | out|typeconversion
```

```python
res=chain.invoke({"language":"Hinglish","query":"Hi how are you?"})
res
```
