### **Langchain basics**

```python
#implementing the chatgroq 
from dotenv import load_dotenv   #api key getsloaded
load_dotenv()
```

```python
from langchain_groq import ChatGroq 
llm = ChatGroq(model="openai/gpt-oss-120b",temperature=0.7) #creating an instance of the chatgroq model
prompt = input("Enter your question:")
res=llm.invoke("Briefly describe about ai")  #invoke the llm instance with a prompt #using langchain work with chatgroq model
res.content
```
