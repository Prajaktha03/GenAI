```python
prompt="Who is pm of india?"
```

```python
import tiktoken
```

```python
tokenizer= tiktoken.encoding_for_model(model_name="gpt-4o")
```

```python
resID=tokenizer.encode(prompt)
```

```python
resID
```

```python
text=tokenizer.decode(resID)
```

```python
text
```
